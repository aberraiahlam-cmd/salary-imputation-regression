## Création des valeurs manquantes

```python
salaire_reel = df['MonthlyIncome'].copy()
```

```python
df_missing = df.copy()
```

```python
np.random.seed(42)
n_missing = int(len(df_missing) * 0.15)
indice_missing = np.random.choice(df_missing.index , size=n_missing, replace=False)
print('Nombre de salaire à masquer:',n_missing)
```

```python
df_missing.loc[indice_missing, 'MonthlyIncome'] = np.nan
```

```python
print('Les valeurs manquantes :', df_missing['MonthlyIncome'].isna().sum())
print('Le pourcentage des valeurs manquantes:', df_missing['MonthlyIncome'].isna().mean() * 100)
```

## Préparation des données

```python
df_train = df_missing[df_missing['MonthlyIncome'].notnull()].copy()
df_test = df_missing[df_missing['MonthlyIncome'].isnull()].copy()
print('Données d entrainement:', df_train.shape)
print('Les données imputer', df_test.shape)
```

```python
feauture = ['Age','TotalWorkingYears','YearsAtCompany','YearsInCurrentRole','YearsWithCurrManager','JobLevel','JobRole',
    'Department','Education','BusinessTravel','MaritalStatus']
target = 'MonthlyIncome'
```

```python
x_train = df_train[feauture]
y_train = df_train[target]
x_test = df_test[feauture]
```

```python
colonnes_num = x_test.select_dtypes(include=['number']).columns
colonnes_cat = x_test.select_dtypes(exclude=['number']).columns
print('Colonnes numériques:', list(colonnes_num))
print('Colonnes catégoriels:', list(colonnes_cat))
```

## Imputation par moyenne et médiane

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler , OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
preprocessing = ColumnTransformer(transformers=[('num', Pipeline([('imputer', SimpleImputer(strategy='median')),('scaler', StandardScaler())]),colonnes_num),
                                  ('cat', Pipeline([('imputer', SimpleImputer(strategy='most_frequent')), ('encoder', OneHotEncoder(handle_unknown='ignore'))])
                                  ,colonnes_cat)])
```

```python
df_mean = df_missing.copy()
impute_mean = SimpleImputer(strategy='mean')
df_mean[['MonthlyIncome']] = impute_mean.fit_transform(df_mean[['MonthlyIncome']])
```

```python
df_median = df_missing.copy()
impute_median = SimpleImputer(strategy='median')
df_median[['MonthlyIncome']] = impute_median.fit_transform(df_median[['MonthlyIncome']])
```

```python
print('Les valeurs manquantes après Imputation par moyenne:', df_mean['MonthlyIncome'].isna().sum())
print('Les valeurs manquantes après Imputation par la medine', df_median['MonthlyIncome'].isna().sum())
```

```python
df_mean['MonthlyIncome']
```

## Imputation par régression linéaire

```python
from sklearn.linear_model import LinearRegression 
model_lr = Pipeline([('preprocessing', preprocessing), ('model', LinearRegression())])
```

```python
model_lr.fit(x_train, y_train)
```

```python
lr_predict = model_lr.predict(x_test)
```

```python
df_lr = df_missing.copy()
df_lr.loc[df_test.index, 'MonthlyIncome'] = lr_predict
```

```python
print(df_lr['MonthlyIncome'].isna().sum())
```

## Imputation par Random Forest

```python
from sklearn.ensemble import RandomForestRegressor
model_rf = Pipeline([('preprocessing', preprocessing), ('model', RandomForestRegressor(n_estimators=100, random_state=42))])
```

```python
model_rf.fit(x_train, y_train)
```

```python
rf_predict = model_rf.predict(x_test)
```

```python
df_rf = df_missing.copy()
df_rf.loc[df_test.index, 'MonthlyIncome'] = rf_predict
```

```python
print(df_rf['MonthlyIncome'].isna().sum())
```

```python
y_test_true = salaire_reel.loc[df_test.index]
```

```python
prediction_mean = np.full(len(df_test), y_train.mean())
prediction_median = np.full(len(df_test), y_train.median())
```

```python
prediction_mean
```

## Évaluation des modèles

```python
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score
def evaluer_model(y_true, y_pred):
    mae = mean_absolute_error(y_true, y_pred)
    mse = mean_squared_error(y_true, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_true, y_pred)
    return [mae, mse, rmse, r2]
```

```python
resultats= { 'moyenne': evaluer_model(y_test_true, prediction_mean),
            'mediane': evaluer_model(y_test_true, prediction_median), 
            'Linear Regression': evaluer_model(y_test_true, lr_predict),
            'Random Forest': evaluer_model(y_test_true,rf_predict)}
df_resultat = pd.DataFrame(resultats, index=['mae', 'mse', 'Rmse', 'r2']).T
print(df_resultat)
```

## Comparaison des résultats

```python
import matplotlib.pyplot as plt 
df_resultat[['mse', 'Rmse']].plot(kind='bar', figsize=(10,5))
plt.title("Comparaison des erreurs des méthodes d'imputation")
plt.ylabel('Erreur')
plt.xticks(rotation=20)
plt.grid(axis='y')
plt.tight_layout()
plt.show()
```

```python
df_resultat['r2'].plot(kind='bar',figsize=(5,7))
plt.title('Comparaison de R2')
plt.xticks(rotation=20)
plt.ylabel('y')
plt.grid(axis='y')
plt.tight_layout()
plt.show()
```

```python
comparaison = pd.DataFrame({'Salaire_reel': y_test_true,
                            'Moyenne': prediction_mean,
                            'Median': prediction_median,
                            'Linear_Regression':lr_predict,
                            'Random_Forest': rf_predict})
print(comparaison.head(10))
```

```python
plt.figure(figsize=(8,6))
plt.scatter(y_test_true, lr_predict, alpha=0.6, label='Linear Regression')
plt.scatter(y_test_true, rf_predict,alpha=0.6,label='Random Forest')
plt.title('Salaires réels vs Salaires prédicts')
plt.xlabel('Salaires réels')
plt.ylabel('Salaire prédicts')
plt.legend()
plt.grid()
plt.tight_layout()
plt.show()
```
